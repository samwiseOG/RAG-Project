```python
import re
prompt = "This is sentence A. This is sentence B. This is sentence C. This is sentence D."
sentences = re.split(r"([.?!])", prompt)
print(sentences)
# Group sentences and their terminators back together
print(sentences[0::2])
print(sentences[1::2])
print(zip(sentences[0::2], sentences[1::2]))
sentences = ["".join(i) for i in zip(sentences[0::2], sentences[1::2])]
print(sentences)
```
